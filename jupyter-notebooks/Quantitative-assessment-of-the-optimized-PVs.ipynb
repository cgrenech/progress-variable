```python
PV_names = [
    'optimized', 
    'PM', 
    'IP', 
    'F', 
    'W', 
    'M1', 
    'M2',
    'X', 
    'K1', 
    'K2', 
    'B' 
]

# Individual costs
costs_all = [
    costs_optimized,
    costs_PM,
    costs_IP,
    costs_F,
    costs_W,
    costs_M1,
    costs_M2,
    costs_X,
    costs_K1,
    costs_K2,
    costs_B
]

# Average/global costs
average_costs = [
    L2_norm_cost_optimized,
    L2_norm_cost_PM,
    L2_norm_cost_IP,
    L2_norm_cost_F,
    L2_norm_cost_W,
    L2_norm_cost_M1,
    L2_norm_cost_M2,
    L2_norm_cost_X,
    L2_norm_cost_K1,
    L2_norm_cost_K2,
    L2_norm_cost_B
]

# Create table with individual costs
table_individual_costs = pd.DataFrame(
    costs_all,
    index=PV_names,
    columns=VD_target_variables_names
)

# Put PV name in the first column
table_individual_costs.index.name = 'PV'

table_individual_costs = table_individual_costs.reset_index()
# Rename 
table_individual_costs = table_individual_costs.rename(
    columns={'PV_ST': '$\\dot{\\omega}_{C}/\\rho$',
             'PV': '${Y}_{C}$',
             'rho': '$\\rho$',
             'mu': '$\\mu$',
             'lambda': '$\\lambda$',
             'c_p':'${c}_{p}$',
             'T': '$T$'
    }
)

# Round numerical values
table_individual_costs.iloc[:, 1:] = (
    table_individual_costs.iloc[:, 1:].round(3)
)

print(table_individual_costs)



def save_cost_table_png(df, filename, fontsize=16):

    fig, ax = plt.subplots(figsize=(22, 6))
    ax.axis('off')

    table = ax.table(
        cellText=df.values,
        colLabels=df.columns,
        cellLoc='center',
        loc='center'
    )

    table.auto_set_font_size(False)
    table.set_fontsize(fontsize)
    table.scale(1, 1.8)

    # Color scale for numerical columns
    for j, column in enumerate(df.columns):

        # Skip first column: PV names
        if j == 0:
            continue

        values = df[column].astype(float).values

        norm = mcolors.Normalize(
            vmin=np.min(values),
            vmax=np.max(values)
        )

        cmap = plt.cm.RdYlGn_r

        for i, value in enumerate(values):
            color = cmap(norm(value))
            table[(i + 1, j)].set_facecolor(color)

    # Header formatting
    for j in range(len(df.columns)):
        table[(0, j)].set_facecolor('lightgray')
        table[(0, j)].set_text_props(weight='bold')

    plt.tight_layout()

    plt.savefig(
        filename,
        dpi=300,
        bbox_inches='tight'
    )

    plt.close()


save_cost_table_png(
    table_individual_costs,
    'costs_comparison.png'
)
table_individual_costs.to_csv("FPF-"+ data_tag +"-costs-species.csv", index=False, header=True)
```
